2. Explain the concept of function arguments and parameters in Python.

  - Arguments are the actual values that you pass to the function when you call it. These values replace the parameters defined in the function.
Example:-

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
  - map():
     Applies a given function to each item in an iterable (e.g., list, tuple) and returns a new iterator with the results. It transforms each element individually.
  - filter():
     Creates a new iterator containing only the items from the original iterable for which a given function returns True. It selectively keeps elements based on a condition.
  - reduce():
     Applies a function cumulatively to the items of an iterable, from left to right, to reduce the iterable to a single value. It aggregates elements.

     ```python
     from functools import reduce

     numbers = [1, 2, 3, 4, 5]

     # map(): double each number
     squared_numbers = list(map(lambda x: x * 2, numbers))  # Output: [2, 4, 6, 8, 10]

     # filter(): keep only even numbers
     even_numbers = list(filter(lambda x: x % 2 == 0, numbers))  # Output: [2, 4]

     # reduce(): calculate the sum of all numbers
     sum_of_numbers = reduce(lambda x, y: x + y, numbers)  # Output: 15
    
     ```

````python
def sum_even(nums):
    return sum(x for x in nums if x % 2 == 0)
````

````python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0: return False
    return True
primes = [n for n in range(1, 201) if is_prime(n)]
````

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

````python
class Fibonacci:
    def __init__(self, n):
        self.n, self.a, self.b = n, 0, 1
        self.count = 0
    def __iter__(self): return self
    def __next__(self):
        if self.count >= self.n: raise StopIteration
        val = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return val
````

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

````python
def powers_of_2(exp):
    for i in range(exp + 1):
        yield 2 ** i
````

7. Implement a generator function that reads a file line by line and yields each line as a string.

````python
def read_lines(filepath):
    with open(filepath) as f:
        for line in f:
            yield line.strip()
````

8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

````python
data = [(1, 3), (2, 1), (4, 2)]
sorted_data = sorted(data, key=lambda x: x[1])
````

````python
celsius = [0, 10, 20]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
````

10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

````python
def remove_vowels(s):
    return ''.join(filter(lambda x: x.lower() not in 'aeiou', s))
````
